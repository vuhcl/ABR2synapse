# Stage 2 HP tuning — true-noise mirror (scenarios A, B, C)

Tunes RF/XGB/MLP under oracle **`noise_cat`** (no Stage 1). Writes `figures/cache/stage2_best_hp_true/{A,B,C}.json`.

- **A** = Brad-only pool
- **B** = combined pool (both cohorts)
- **C** = Liberman-only pool

Run **before** [`abr_stage2_synthesis_cv_true_noise.ipynb`](abr_stage2_synthesis_cv_true_noise.ipynb).

Predicted vs true HP are **not interchangeable**.

```python
from pathlib import Path
import pandas as pd

from utils.nn_stage2_data import load_nn_stage2_data
from utils.stage2_export import export_stage2_data
from utils.stage2_hp import (
    resolve_torch_device,
    tune_scenario,
    write_scenario_hp_json,
)
from utils.benchmark_metrics import STAGE2_BEST_HP_TRUE_DIR, STAGE2_DATA_DIR

HP_DIR = STAGE2_BEST_HP_TRUE_DIR
HP_DIR.mkdir(parents=True, exist_ok=True)
device = resolve_torch_device()
print("device:", device)
```

```python
if not (STAGE2_DATA_DIR / "manifest.json").is_file():
    export_stage2_data(out_dir=STAGE2_DATA_DIR)
else:
    print("stage2_data export present; skip")
```

```python
data = load_nn_stage2_data()
trials = []
for scen in ("A", "B", "C"):
    payload, mlp_log = tune_scenario(
        data, scen, noise_label="true", device=device, verbose=True
    )
    write_scenario_hp_json(payload, HP_DIR / f"{scen}.json")
    trials.append(mlp_log)
pd.concat(trials, ignore_index=True).to_parquet(
    HP_DIR / "hp_search_log_true.parquet", index=False
)
print("Wrote", HP_DIR)
```
